```python
import pandas as pd
```

```python
awards_2025 = pd.read_csv("../data/clean/last_year_tracks.csv", sep=';')
```

**Most followed artist - 2025**

```python
#Pivot table for followers visualisation
most_followed_artist = awards_2025.pivot_table(
    index="artist_name",
    values=["artist_followers"],
    aggfunc={"artist_followers": "max"}
).sort_values(by="artist_followers", ascending=False).head(10)

display(most_followed_artist)
```

```python
#Preparing pivot table for printing - followers
top_3_followers = most_followed_artist.head(3).reset_index()

artist_1, followers_1 = top_3_followers.loc[0, "artist_name"], top_3_followers.loc[0, "artist_followers"]
artist_2, followers_2 = top_3_followers.loc[1, "artist_name"], top_3_followers.loc[1, "artist_followers"]
artist_3, followers_3 = top_3_followers.loc[2, "artist_name"], top_3_followers.loc[2, "artist_followers"]
```

```python
print("Most followed artists in 2025:")
print(f"Top 1 - {artist_1} with {followers_1:,} followers")
print(f"Top 2 - {artist_2} with {followers_2:,} followers")
print(f"Top 3 - {artist_3} with {followers_3:,} followers")
```

**Genre awards preparation**

```python
#Genre function mapping for main categories
analysis_df = awards_2025.copy()

def map_genre(genre_str):
    if pd.isna(genre_str):
        return "Unspecified / Missing"

    subgenres = [g.strip().lower() for g in genre_str.split(",")]

    pop_keywords = ["pop", "synthpop", "art pop", "k-pop", "c-pop", "dance pop", "soft pop", "europop", "teen pop"]
    hiphop_keywords = ["hip hop", "hip-hop", "rap", "trap", "drill", "phonk", "melodic rap", "mahraganat", "freestyle", "boom bap", "boombap"]
    rock_keywords = ["rock", "metal", "indie", "alternative", "psychedelic", "punk", "grunge", "shoegaze", "britpop", "post-rock"]
    rnb_keywords = ["r&b", "rnb", "soul", "funk", "neo soul", "quiet storm"]
    electronic_keywords = ["house", "edm", "techno", "electro", "dance", "club", "trance", "bass", "garage", "hypertechno", "hardstyle", "dubstep"]
    country_keywords = ["country", "folk", "americana", "bluegrass"]

    for sub in subgenres:
        if any(kw in sub for kw in pop_keywords):
            return "Pop"
        if any(kw in sub for kw in hiphop_keywords):
            return "HipHop"
        if any(kw in sub for kw in rock_keywords):
            return "Rock"
        if any(kw in sub for kw in rnb_keywords):
            return "R&B"
        if any(kw in sub for kw in electronic_keywords):
            return "Electronic Music"
        if any(kw in sub for kw in country_keywords):
            return "Country"
            
    return "Other"
```

```python
analysis_df["main_genre"] = analysis_df["artist_genres"].apply(map_genre)
print(analysis_df["main_genre"].value_counts())
```

```python
#Pop Pivot Table
pop_data = analysis_df[analysis_df["main_genre"] == "Pop"]

pivot_pop = pop_data.pivot_table(
    index=["artist_name"],
    values=["track_popularity"],
    aggfunc={"track_popularity": "mean"}
).round(0).astype(int).sort_values(by="track_popularity", ascending=False).head(10)

display(pivot_pop)
```
