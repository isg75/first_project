First_project_day1_Bindu: Deal with the missing values 

```python
import pandas as pd
```

```python
url = r"../data/raw/imdb_top_1000.csv"
df = pd.read_csv(url)
df.head()
```

```python
#Missing values
df.isna().sum()
# 100 out of 1000 certificate values are null
```

```python
#to know data types of each column
df.dtypes
```

```python
# gives the different certificate with count, including NaN
df["Certificate"].value_counts(dropna=False)  
```

```python
# Replace NaN values with Unrated
df["Certificate"]= df["Certificate"].fillna("Unrated")
```

```python
# ratings from India, US and TV are grouped into 4 audience levels. 

age_groups = {"U": "All ages",
              "G": "All ages",
              "GP": "All ages",
              "Passed": "All ages",
              "Approved": "All ages",
              "UA": "Family/Teen",
              "U/A": "Family/Teen",
              "PG-13": "Family/Teen",
              "PG": "Family/Teen",
              "GP": "Family/Teen",
              "TV-PG": "Family/Teen",
              "TV-14": "Family/Teen",
              "A": "Adult",
              "R": "Adult",
              "TV-MA": "Adult",
              "16": "Adult",
              "Unrated": "Unrated"    
    }
df["Audience"] = df["Certificate"].map(age_groups)
display(df["Audience"])
```

```python
df["Audience"].value_counts() # ratings from India, US and TV are grouped into 4 audience levels. 
```

```python
df["Audience"].isna().sum() # no NaN values
```

```python
df["Meta_score"].unique() 
```

```python
df["Main_Genre"]=df["Genre"].str.split(",").str[0].str.strip()
df["Main_Genre"].unique().tolist()
```

```python
Meta_score_medians = df.groupby ("Main_Genre") ["Meta_score"].median()
Meta_score_medians 
```

```python
# Saving the NaN meta score in a df
df["Meta_missing"] = df["Meta_score"].isna()

# Filling the NaN meta score with the medians grouped by genre
df["Meta_score"] = df["Meta_score"].fillna(df["Main_Genre"].map(Meta_score_medians)) 

df["Meta_score"].unique()
```

```python
#checking the missing values with the Main_genre "Fantasy"
df[(df["Meta_score"].isna()) & (df["Main_Genre"]== "Fantasy")]
```

```python
# Filling the remaining NaN values with the overall median
df["Meta_score"] = df["Meta_score"].fillna(df["Meta_score"].median())
```
