```python
import pandas as pd
import os

print(os.listdir("data"))
print(os.listdir("data/raw"))
```

```python
airbnb = pd.read_csv("data/raw/airnb.csv")
```

```python
airbnb.head()
```

```python
airbnb["Title"].head(20)
```

```python
airbnb[["BnB_Type", "Location"]] = airbnb["Title"].str.split(
    " in ",
    n=1,
    expand=True
)
```

```python
airbnb[["City", "Country"]] = airbnb["Location"].str.split(
    ",",
    n=1,
    expand=True
)

# Remove extra spaces
airbnb["City"] = airbnb["City"].str.strip()
airbnb["Country"] = airbnb["Country"].str.strip()
```

```python
split = airbnb["Title"].str.split(" in ", n=1, expand=True)

airbnb["BnB_Type"] = split[0]
airbnb["Location"] = split[1]
airbnb[["BnB_Type", "Location"]].isnull().sum()
```

```python
# Split Location into City and Country
airbnb[["City", "Country"]] = airbnb["Location"].str.split(
    ",",
    n=1,
    expand=True
)

# Remove leading and trailing spaces
airbnb["City"] = airbnb["City"].str.strip()
airbnb["Country"] = airbnb["Country"].str.strip()
```

```python
airbnb[["Location", "City", "Country"]].head(10)
```

```python
# Extract the first month abbreviation
airbnb["Month"] = airbnb["Date"].str.extract(r"^([A-Za-z]{3})")[0]

# Convert to full month name
month_map = {
    "Jan": "January",
    "Feb": "February",
    "Mar": "March",
    "Apr": "April",
    "May": "May",
    "Jun": "June",
    "Jul": "July",
    "Aug": "August",
    "Sep": "September",
    "Oct": "October",
    "Nov": "November",
    "Dec": "December"
}

airbnb["Month"] = airbnb["Month"].map(month_map)

airbnb[["Date", "Month"]].head()
```

```python
# Rating
airbnb["Rating"] = airbnb["Review and rating"].str.extract(r"(\d+\.\d+)")

# Review count
airbnb["Reviews"] = airbnb["Review and rating"].str.extract(r"(\d+)")

# Convert to numeric
airbnb["Rating"] = pd.to_numeric(airbnb["Rating"], errors="coerce")
airbnb["Reviews"] = pd.to_numeric(airbnb["Reviews"], errors="coerce")
```

```python
airbnb["Review and rating"].head(10)
```

```python
# Extract Rating
airbnb["Rating"] = airbnb["Review and rating"].str.extract(r'(\d+\.\d+|\d+)')

# Extract Number of Reviews
airbnb["Reviews"] = airbnb["Review and rating"].str.extract(r'\((\d+)\)')

# Convert both columns to numeric
airbnb["Rating"] = pd.to_numeric(airbnb["Rating"])
airbnb["Reviews"] = pd.to_numeric(airbnb["Reviews"])

# Display the result
airbnb[["Review and rating", "Rating", "Reviews"]].head(10)
```

```python
# Get the list of columns
cols = airbnb.columns.tolist()

# Remove the Review and Rating columns
cols.remove("Reviews")
cols.remove("Rating")

# Find the position of the Detail column
detail_index = cols.index("Detail")

# Insert Rating and Reviews after Detail
cols.insert(detail_index + 1, "Rating")
cols.insert(detail_index + 2, "Reviews")

# Reorder the DataFrame
airbnb = airbnb[cols]

# Display the first few rows
airbnb.head()
```

```python
# Remove rows with missing values in Country, Rating, or Reviews
airbnb = airbnb.dropna(subset=["Country", "Rating", "Reviews"])

# Reset the index
airbnb.reset_index(drop=True, inplace=True)

# Check the result
airbnb.info()
```

```python
import pandas as pd

# Replace blank strings with NaN
airbnb[["Country", "Rating", "Reviews"]] = (
    airbnb[["Country", "Rating", "Reviews"]]
    .replace(r'^\s*$', pd.NA, regex=True)
)

# Remove rows with missing values
airbnb = airbnb.dropna(subset=["Country", "Rating", "Reviews"])

# Reset the index
airbnb.reset_index(drop=True, inplace=True)
```

```python
# Rename columns
airbnb.rename(columns={
    "Title": "Property Name",
    "Detail": "Amenities",
    "BnB_Type": "Property Type",
    "Reviews": "Number of Reviews"
}, inplace=True)

# Check the updated columns
airbnb.columns.tolist()
```

```python
airbnb = airbnb.sort_values(
    by=["Rating", "Number of Reviews"],
    ascending=[False, False]
)

airbnb["Rating Rank"] = range(1, len(airbnb) + 1)

airbnb.head(20)
```

```python
# Create Price Rank (Highest Price = Rank 1)
airbnb["Price Rank"] = airbnb["Price(in dollar)"].rank(
    ascending=False,
    method="dense"
).astype(int)

# Sort from highest price to lowest
airbnb = airbnb.sort_values(
    by="Price(in dollar)",
    ascending=False
)

# Display the result
airbnb.head(20)
```
