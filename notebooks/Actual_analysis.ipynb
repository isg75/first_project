```python
# 2. Country Analysis

This section examines the distribution of Airbnb properties across countries and compares their average prices and ratings.
```

```python
query = """
SELECT
    Country,
    COUNT(*) AS total_properties
FROM airbnb_listings
GROUP BY Country
ORDER BY total_properties DESC;
"""

country_count = pd.read_sql(query, engine)

country_count
```

```python
query = """
SELECT
    Country,
    COUNT(*) AS total_properties,
    ROUND(
        COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (),
        2
    ) AS percentage
FROM airbnb_listings
GROUP BY Country
ORDER BY total_properties DESC;
"""

country_count = pd.read_sql(query, engine)

country_count
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=country_count,
    x="total_properties",
    y="Country"
)

plt.title("Number of Properties by Country")
plt.xlabel("Number of Properties")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
query = """
SELECT
    Country,
    COUNT(*) AS total_properties,
    ROUND(AVG(`Price(in dollar)`), 2) AS average_price
FROM airbnb_listings
GROUP BY Country
ORDER BY average_price DESC;
"""

country_price = pd.read_sql(query, engine)

country_price
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=country_price,
    x="average_price",
    y="Country"
)

plt.title("Average Property Price by Country")
plt.xlabel("Average Price in USD")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
query = """
SELECT
    Country,
    COUNT(*) AS total_properties,
    ROUND(AVG(Rating), 2) AS average_rating
FROM airbnb_listings
GROUP BY Country
ORDER BY average_rating DESC;
"""

country_rating = pd.read_sql(query, engine)

country_rating
```

```python
plt.figure(figsize=(12, 7))

sns.barplot(
    data=country_rating,
    x="average_rating",
    y="Country"
)

plt.title("Average Property Rating by Country")
plt.xlabel("Average Rating")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

```python
query = """
SELECT
    Country,
    COUNT(*) AS total_properties,
    ROUND(AVG(`Price(in dollar)`), 2) AS average_price,
    ROUND(AVG(Rating), 2) AS average_rating,
    SUM(`Number of Reviews`) AS total_reviews
FROM airbnb_listings
GROUP BY Country
ORDER BY total_properties DESC;
"""

country_summary = pd.read_sql(query, engine)

country_summary
```
