### Comparative analysis of the two CSV files

* Are the Excels the same data? Or different data?
    * Columns, names, contents, and data types
    * Merge tables


```python
import pandas as pd

# 1. Loading the two CSV files
df1 = pd.read_csv('../data/raw/spotify_data clean.csv')
df2 = pd.read_csv('../data/raw/track_data_final.csv')

print(f"df1 shape: {df1.shape}")
print(f"df2 shape: {df2.shape}")
```

```python

# 2. Checking the basic structure (matching columns)
cols_df1 = list(df1.columns)
cols_df2 = list(df2.columns)



if cols_df1 == cols_df2:
    print("The names and order of the columns are the same")
else:
    print("There is a discrepancy between the columns!")
    print("columns of df1:", list(df1.columns))
    print("\ncolumns of df2:", list(df2.columns))

# 3. Comparison of full content (can also be examined regardless of order)
# The equals() method checks for an exact match (it assumes that the order is also the same)

if df1.equals(df2):
    print("\nThe two dataframe are the same.")
else:
    print("\nThere is a difference between the contents of the two files.")
    

print("-" * 60)

# 3. Comparison of Data Types
dtypes_df = pd.DataFrame({
    'df1_types': df1.dtypes,
    'df2_types': df2.dtypes
})
dtypes_df['match'] = dtypes_df['df1_types'] == dtypes_df['df2_types']
print("--- Comparison of Data Types ---")
print(dtypes_df)

print("-" * 60)

# 4. Checking for exact match of the entire content
if df1.equals(df2):
    print("The content and structure of the two DataFrames are identical in every respect.")
else:
    print("The contents differ.")

print("-" * 60)
```

```python
# 5. We check how many common track_ids exist between the two dataframes
# The .isin() method checks whether the track_ids in df1 are present in df2
shared_songs = df1['track_id'].isin(df2['track_id'])

# The number of shared songs
shared_songs_nr = shared_songs.sum()
print(f"Number of songs common to (duplicated in) the two files: {shared_songs_nr}")
print(f"df1 contains a total of {len(df1)} songs, while df2 contains {len(df2)} songs.")

print("-" * 60)

# 6. print the actual rows that appear in both places:
# Filter df1 for the common identifiers
duplicated_rows_df1 = df1[shared_songs]

# print("--- Example of songs present in both files (based on df1) ---")
# display(duplicated_rows_df1.head(10))

print("-" * 60)

# 7. Check in the opposite direction as well:
# Are there any songs in df2 that are not in df1?
only_in_df2 = df2[~df2['track_id'].isin(df1['track_id'])]
print(f"Number of (unique) songs present only in the 2nd file: {len(only_in_df2)}")
```

```python
# 8. Check if all track_ids of df1 are present in df2
is_subset = set(df1['track_id']).issubset(set(df2['track_id']))
print(f"Is df1 a complete subset of df2?: {is_subset}")

print("-" * 60)

# 9. Find and display the extra rows that are in df2 but NOT in df1
extra_rows_in_df2 = df2[~df2['track_id'].isin(df1['track_id'])]

print(f"Number of extra rows only in df2: {len(extra_rows_in_df2)}")
if len(extra_rows_in_df2) > 0:
    print("\n--- Preview of extra rows found only in df2 ---")
    display(extra_rows_in_df2.head(10))
```

```python
print(f"Is track_id unique in df1?: {df1['track_id'].is_unique}")
print(f"Is track_id unique in df2?: {df2['track_id'].is_unique}")

print("-" * 60)

# 10. Check for fully duplicated rows (all columns match) in df2
full_duplicates_df2 = df2[df2.duplicated(keep=False)]
print(f"Number of fully duplicated rows in df2: {len(full_duplicates_df2)}")

if len(full_duplicates_df2) > 0:
    print("\n--- Preview of fully duplicated rows in df2 ---")
    display(full_duplicates_df2.head(10))
else:
    print("No fully duplicated rows found in df2.")
```
