```python
import pandas as pd

def title_extractor(df: pd.DataFrame) -> pd.DataFrame:
    temp_df = df.copy()
    temp_df[["header", "state", "country"]] = temp_df["title"].str.split(
        ",", expand=True
    )
    temp_df.drop("title", inplace=True, axis=1)
    return temp_df
```

Finding location of every listing: We can use 'in' to get the location

```python
def header_extractor(df: pd.DataFrame) -> pd.DataFrame:
    temp_df = df.copy()

    # Split header into type and location
    split_header = temp_df["header"].str.split(" in ", n=1, expand=True)

    # Give the two resulting columns names
    split_header.columns = ["type", "location"]

    # Add them to the dataframe
    temp_df[["type", "location"]] = split_header

    # Remove original header
    temp_df.drop(columns=["header"], inplace=True)

    return temp_df
```

```python
from geopy.geocoders import Nominatim
from time import sleep

geolocator = Nominatim(
    user_agent="airbnb_analysis_project"
)
```

```python
def get_coordinates(row):

    address = f"{row['location']}, {row['state']}, {row['country']}"

    try:
        location = geolocator.geocode(
            address,
            timeout=10
        )

        if location:
            return pd.Series({
                "latitude": location.latitude,
                "longitude": location.longitude
            })

    except Exception as e:
        print(f"Error for {address}: {e}")

    return pd.Series({
        "latitude": np.nan,
        "longitude": np.nan
    })
```

Iterating to fill information of latitude and longitude in current data frame

```python
for index, row in df_location_us.iterrows():

    print(
        index,
        row["location"],
        row["state"]
    )

    coordinates = get_coordinates(row)

    df_location_us.loc[
        index, "latitude"
    ] = coordinates["latitude"]

    df_location_us.loc[
        index, "longitude"
    ] = coordinates["longitude"]

    sleep(1)
```
