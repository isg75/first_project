```python
import pandas as pd

def title_extractor(df: pd.DataFrame) -> pd.DataFrame:
    temp_df = df.copy()
    temp_df[["Header", "State", "Country"]] = temp_df["Title"].str.split(
        ",", expand=True
    )
    temp_df.drop("Title", inplace=True, axis=1)
    return temp_df
```

```python
def header_extractor(df: pd.DataFrame) -> pd.DataFrame:
    temp_df = df.copy()

    # Split header into type and location
    split_header = temp_df["header"].str.split(" in ", n=1, expand=True)

    # Give the two resulting columns names
    split_header.columns = ["type", "location"]

    # Add them to the dataframe
    temp_df[["type", "location"]] = split_header

    # Remove original header
    temp_df.drop(columns=["header"], inplace=True)

    return temp_df
```

```python
import pandas as pd
from geopy.geocoders import Nominatim
from IPython.display import displayimport numpy as np # Import numpy for np.nan

# Ensure 'location', 'state', and 'country' columns are strings for concatenation
# This also handles potential 'nan' float values by converting them to string 'nan'
df_location_us['location'] = df_location_us['location'].astype(str)
df_location_us['state'] = df_location_us['state'].astype(str)
df_location_us['country'] = df_location_us['country'].astype(str)

# Create or update the 'search_location' column for df_location_us
# Replacing 'nan' string with an empty string to avoid incorrect geocoding strings
df_location_us["search_location"] = (
    df_location_us["location"].replace('nan', '') + ", " +
    df_location_us["state"].replace('nan', '') + ", " +
    df_location_us["country"].replace('nan', '')
)

# Clean up any extra commas or spaces that might result from concatenation
df_location_us["search_location"] = df_location_us["search_location"].str.replace(r',\s*$', '', regex=True).str.strip()
df_location_us["search_location"] = df_location_us["search_location"].str.replace(r'^,\s*', '', regex=True).str.strip()

# Define the geocoding function (re-introducing it as it was deleted)
geolocator = Nominatim(user_agent="airbnb_analysis_colab_project", timeout=10)

def get_coordinates(place):
    if pd.isna(place) or str(place).strip() == '': # Handle NaN or empty strings
        return np.nan, np.nan
    try:
        location = geolocator.geocode(place, timeout=10)
        if location:
            return location.latitude, location.longitude
    except Exception as e:
        # Print a message for debugging, but don't stop the process
        print(f"Error geocoding '{place}': {e}")
    return np.nan, np.nan

# Apply the geocoding function to populate latitude and longitude
print("Geocoding locations for df_location_us...")
df_location_us[['latitude', 'longitude']] = df_location_us['search_location'].apply(
    lambda x: pd.Series(get_coordinates(x))
)

print("\nGeocoding complete. Displaying first 5 rows with new coordinates:")
display(df_location_us[['search_location', 'latitude', 'longitude']].head())

# Filter out rows where geocoding failed (latitude or longitude is NaN)
initial_rows = df_location_us.shape[0]
df_location_us.dropna(subset=['latitude', 'longitude'], inplace=True)
rows_after_geocoding_drop = df_location_us.shape[0]
print(f"Dropped {initial_rows - rows_after_geocoding_drop} rows due to failed geocoding.")

print("\nDataFrame info after geocoding and dropping null coordinates:")
df_location_us.info()
```
