```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/track_data_final.csv")
df.head()
```

```python
df.columns.tolist()
```

```python
df_reduced = df.drop(columns=["track_number","explicit"])
df_reduced.head()
```

```python
df_reduced.nunique()
```

```python
df_no_genre = df_reduced[df_reduced["artist_genres"] == "[]"]
display(df_no_genre)
display(df_no_genre.info())
```

```python
df_no_genre[df_no_genre["track_name"].isna()]
```

```python
df_no_genre = df_no_genre.dropna()
df_no_genre.shape
```

```python
df_no_genre["album_year"] = df_no_genre["album_release_date"].str[:4]
```

```python
df_no_genre["album_year"] = df_no_genre['album_year'].astype(int)
```

```python
df_no_genre_yr_filter = df_no_genre[df_no_genre["album_year"] > 2015]
```

```python
df_no_genre_yr_filter.shape
```

```python
artist_album_no_genre = (
    df_no_genre_yr_filter.
        groupby(["artist_name", "album_name"])
        .size()
    .sort_values(ascending=False)
)
display(artist_album_no_genre.head(50))
display(artist_album_no_genre.size)
```

```python
artist_no_genre = (
    df_no_genre_yr_filter.
        groupby(["artist_name"])
        .size()
    .sort_values(ascending=False)
)
display(artist_no_genre.head(50))
print(artist_no_genre.shape)

print("\nDetermining the cut_off for majority of data")
print(artist_no_genre.describe())

print("\nTotal number of ")
```

```python
print("\nDetermining the value counts in descending order: ")
print(artist_no_genre.value_counts(ascending=True))
print("\nTotal check of counts: ")
print(sum(artist_no_genre.value_counts(ascending=True)))
```

```python
artist_no_genre_cutoff = artist_no_genre[artist_no_genre > 2]
display(artist_no_genre_cutoff.value_counts(ascending=True))
print(sum(artist_no_genre_cutoff.value_counts(ascending=True)))
```

```python
pd.set_option('display.max_rows', 250)
display(artist_no_genre_cutoff)
```

```python
artist_need_genre_label = artist_no_genre_cutoff.index.tolist()
```

```python
df_artist_need_genre_label = (
    artist_no_genre_cutoff
    .to_frame()
    .reset_index()
)
```

```python
df_artist_need_genre_label.to_csv("artist_genre_mapping.csv", index=False)
```
