## Cleaning pricein_dollar and title columns

### formatting title

```python
# "type in location", for example '(type:)Apartment in (location:)Tulum, Mexico'
# n=1 because i want just to separate from the first "in" and not in another possible others 'in' alongside
airbnb_df[["property_type", "location"]] = airbnb_df["title"].str.split(" in ", n=1, expand=True) 
airbnb_df.head()
```

#### in the dataset, 'location' can be only city, or city+country, or city+state+country
#### we need to know how many ' , ' exists to confirm which formats they are

```python
airbnb_df["location"].str.count(",").value_counts()
```

#554 'city, country'

#240 with just 'city'

#125 'city, state, country'

```python
location_parts = airbnb_df["location"].str.split(",", expand=True)
location_parts.head()
```

#splitting the 'location' column where there is a ' , ' , to have later separate city state and country columns based on position

```python
location_parts[2].isna().sum()
```

#checking how many rows have a value in index [2] after the split

#expected: 794 nulls (919 - 125), since only 125 rows had 2 commas

#city + state + country format

```python
# to remove any space 

location_parts = location_parts.apply(lambda col: col.str.strip())
```

```python
# city is always the first split value (index 0) 
# + every row has at least a city name
# + confirmed earlier that column 0 has no missing values

airbnb_df["city"] = location_parts[0]
```

```python
# country = column 2 when it exists (city+state+country format); 
# otherwise -> falls back to column 1 (city+country format, no state)

airbnb_df["country"] = location_parts[2].where(location_parts[2].notna(), location_parts[1])
```

```python
# state only exists when there was a 3rd part (column 2 present)
# in the city+country rows, column 1 is already used as country 
# so state is correctly left as NaN there
airbnb_df["state"] = location_parts[1].where(location_parts[2].notna())
```

```python
location_parts[location_parts[2].notna()].head()
```

```python
airbnb_df[["city", "state", "country"]].isna().sum()
```

# Establishing relation between price and property_type

#### Average price by property_type

```python
# reliable vs low-volume property types
price_by_type = (airbnb_df.groupby("property_type")["price"].agg(["count", "mean", "median"]).round(2).sort_values("mean", ascending=False))
price_by_type
```

```python
# property types with count >= 10 , so we can trust the listings in order to trust the average
reliable_types = price_by_type[price_by_type["count"] >= 10]
reliable_types
```

```python
# property types with count < 10
# just one or two listings = these shouldn't be read as reliable patterns
low_volume_types = price_by_type[price_by_type["count"] < 10]
low_volume_types
```

#### insights:

#### - Villa has the highest average price (~287$),
#### however Home have just a smaller average price (~242$) the double amount of listings (114 > 59). In that sense, the amount of money made by the type Home seems to be greater than the one made by Villa type, bc it represents the largest share of the market by listing count;

#### - Higher volume =/= higher price, bc:
####   -- Cabin accounts for 115 listings (the biggest in the dataset) but with an average of ~193$,
####   -- Apartment with 115 listings only with an average of 125$,
####   -- Condo with 88 listings and average of 168$ 

#### - Mean vs median: cottage, and bungalow have greater disparity between mean and median, which means that the outliers in the list are pulling the average up, so this is worth taking in account for possible price adjustments(?)

```python
price_by_type["estimated_total"] = price_by_type["mean"] * price_by_type["count"]
price_by_type.sort_values("estimated_total", ascending=False)
```
