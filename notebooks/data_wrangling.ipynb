```python
# Find intersection of job_ids
common_ids = set(df_data_analyst['job_id']).intersection(set(df_keywords['job_id']))

# Extract rows from either DataFrame with those ids
df_overlap_analyst = df_data_analyst[df_data_analyst['job_id'].isin(common_ids)]
df_overlap_keywords = df_keywords[df_keywords['job_id'].isin(common_ids)]

df_overlap = df_merged[df_merged['job_id'].isin(common_ids)]

# Show all overlapping job_ids
print("Number of overlapping jobs:", len(common_ids))
print("Overlapping job_ids:", common_ids)

# Show the rows
df_overlap.head(10)
```

```python
# Create the exclusion key set
exclude_keys = pd.concat([df_data_analyst[['job_id']], df_keywords[['job_id']]]).drop_duplicates()

# Merge with indicator to track exclusive rows
df_excluded = df_merged.merge(exclude_keys, on=['job_id'], how='left', indicator=True)

# Keep only rows that were NOT matched
df_other = df_excluded[df_excluded['_merge'] == 'left_only'].drop(columns=['_merge'])

df_excluded['_merge'].unique()
```

```python
# Audit Explanation:
# Naive sum double-counts overlaps, so it’s larger than the universe (df_merged).
# Subtracting overlap gives the original postings count, confirming your splits are accurate.
# This is normal and expected in set operations where groups overlap.
summary = pd.DataFrame({
    'DataFrame': ['df_merged', 'df_data_analyst', 'df_keywords', 'df_other', 'df_overlap', 'Sum (naive)', 'Sum (minus overlap)'],
    'Count': [len(df_merged), len(df_data_analyst), len(df_keywords), len(df_other), len(df_overlap), 
              len(df_data_analyst) + len(df_keywords) + len(df_other), 
              len(df_data_analyst) + len(df_keywords) + len(df_other) - len(df_overlap)],
    'Description': [
        'All unique job postings',
        'Data Analyst/Engineer job postings',
        'Mentions data skills in preferred_skills',
        'Remaining postings (not analyst/keywords)',
        'Postings in both analyst and keywords',
        'Naive sum (overlap double-counted)', 
        'Corrected sum (overlap subtracted)'
    ]
})

display(summary)
```
