```python
# Standardise column names: lowercase, snake_case
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# df = df.rename(columns={"": ""})

# remove: "poster_link" + "overview"
columns_to_keep = ["series_title", "released_year", "certificate", "runtime", "genre",
"imdb_rating", "meta_score", "director", "star1", "star2", "star3", "star4",
"no_of_votes", "gross"]

df_clean = df[columns_to_keep].copy()
df_clean
```

```python
# convert released year from string to int
df_clean["released_year"] = df_clean["released_year"].astype(int)
```

```python
# convert runtime from string to int
df_clean["runtime"] = df_clean["runtime"].str.replace(" min", "", regex=False).astype(int)
```
