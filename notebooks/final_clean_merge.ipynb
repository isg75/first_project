```python
# Import data for the country missing it on gdp and pop:
car_df=pd.read_csv("../data/raw/CAR_gdp_pop.csv")
car_df.head()
```

```python
# Cleaning df and preparing it to fill the missing data:
car_df.columns = [str(col).split('[')[0].strip().lower().replace(' ', '_')for col in car_df.columns]
car_df=car_df.drop(columns=['series_code','country_code'])
car_df = car_df.rename(columns={'country_name': 'country'})
car_df.columns
```

```python
# getting rid of unnecessary rows:
car_df=car_df.dropna()
car_df.isnull()
```

```python
# melt + pivot the data to match the destination format:
years=['2014','2015','2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024']
car_melt = pd.melt(car_df, id_vars=['country', 'series_name'], value_vars=years, var_name='year', value_name='value')
display(car_melt.head())

car_final=car_melt.pivot(index=['country','year'],columns='series_name',values='value').reset_index()
display(car_final.head())
car_final.dtypes
```

```python
# Match column names to the energy df and change year to integer:
car_final = car_final.rename(columns={'GDP (current US$)': 'GDP','GDP per capita (current US$)': 'gdp_per_cap','Population, total': 'population'})
display(car_final.head(12))
```

```python
#Add the new data to our final dataset:
car_final['year']=car_final['year'].astype(int)

final_df = total.merge(car_final,on=['country', 'year'],how='left')
display(final_df.columns)
```

```python
#Filling missing data
final_df['GDP_x'] = final_df['GDP_x'].fillna(final_df['GDP_y'])
final_df['gdp_per_cap_x'] = final_df['gdp_per_cap_x'].fillna(final_df['gdp_per_cap_y'])
final_df['population_x'] = final_df['population_x'].fillna(final_df['population_y'])

final_df = final_df.drop(columns=['GDP_y', 'gdp_per_cap_y', 'population_y'])

final_df = final_df.rename(columns={'GDP_x': 'GDP','gdp_per_cap_x': 'gdp_per_cap','population_x': 'population'})
```
