```python
import pandas as pd

url = "https://www.sharkattackfile.net/spreadsheets/GSAF5.xls"

df = pd.read_excel(url)

# Preview the data
print(df.head())

# Check dimensions
print(df.shape)

# View column names
display(df.columns.tolist())
```

```python
df.info()
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
columns = [
    "Date",
    "Year",
    "Type",
    "Country",
    "Location",
    "Activity",
    "Sex",
    "Age",
    "Injury",
    "Time"
]

df = df[columns]
display (df)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("/", "_", regex=False)
    .str.replace(r"[^a-z0-9_]", "", regex=True)
)
display (df)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("/", "_", regex=False)
    .str.replace("(", "", regex=False)
    .str.replace(")", "", regex=False)
)
display(df.head())
```

```python
print(df.columns.tolist())
```

```python
columns_needed = [
    "date",
    "year",
    "type",
    "country",
    "location",
    "activity",
    "sex",
    "age",
    "injury",
    "time"
]

df_clean = df[columns_needed].copy()
display(df_clean.head())
```

```python
display(df_clean["sex"].value_counts(dropna=False))
display(df_clean["type"].value_counts(dropna=False))
display(df_clean["country"].value_counts(dropna=False))
```

```python
text_columns = [
    "type",
    "country",
    "location",
    "activity",
    "sex",
    "injury",
    "time"
]

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()
    display(df_clean[column].value_counts(dropna=False))
```

```python
df_clean["sex"] = df_clean["sex"].str.upper().str.strip()

sex_replacements = {
    "M ": "M",
    "F ": "F",
    "MALE": "M",
    "FEMALE": "F"
}

df_clean["sex"] = df_clean["sex"].replace(sex_replacements)
display(df_clean["sex"].value_counts(dropna=False))
```

```python
valid_sex = ["M", "F"]

invalid_sex = (
    df_clean["sex"].notna()
    & ~df_clean["sex"].isin(valid_sex)
)

display(df_clean.loc[invalid_sex, "sex"].value_counts())
```

```python
df_clean["type"] = df_clean["type"].str.title().str.strip()

display(df_clean["type"].value_counts(dropna=False))
```

```python
df_clean["country"] = (
    df_clean["country"]
    .str.upper()
    .str.strip()
)

country_replacements = {
    "USA": "UNITED STATES",
    "U.S.A.": "UNITED STATES",
    "UNITED STATES OF AMERICA": "UNITED STATES",
    "ENGLAND": "UNITED KINGDOM"
}

df_clean["country"] = df_clean["country"].replace(country_replacements)
display(df_clean["country"].value_counts(dropna=False))
```

```python
df_clean["activity"] = (
    df_clean["activity"]
    .str.lower()
    .str.strip()
)

display(df_clean["activity"].value_counts().head(30))
```

```python
def clean_activity(value):
    if pd.isna(value):
        return value

    value = value.lower().strip()

    if "surf" in value:
        return "surfing"
    elif "swim" in value:
        return "swimming"
    elif "div" in value:
        return "diving"
    elif "fish" in value:
        return "fishing"
    elif "snorkel" in value:
        return "snorkeling"
    elif "wading" in value:
        return "wading"
    else:
        return value

df_clean["activity_clean"] = df_clean["activity"].apply(clean_activity)
display(df_clean["activity_clean"].value_counts().head(30))
```

```python
display(df_clean["age"].value_counts(dropna=False).head(30))
```

```python
age_text = df_clean["age"].astype("string").str.strip()

invalid_age = (
    age_text.notna()
    & ~age_text.str.fullmatch(r"\d+")
)

display(age_text[invalid_age].value_counts().head(30))
```

```python
null_counts = df.isnull().sum()
null_counts = null_counts[null_counts > 0]

display(null_counts)
```

```python
null_summary = pd.DataFrame({
    "null_count": df.isnull().sum(),
    "null_percentage": df.isnull().mean() * 100
})

null_summary = null_summary[null_summary["null_count"] > 0]
null_summary = null_summary.sort_values(
    by="null_percentage",
    ascending=False
)

display(null_summary)
```

```python
categorical_columns = [
    "type",
    "country",
    "location",
    "activity",
    "name",
    "sex",
    "injury",
    "time"
]

for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna("Unknown")
        display(df[col].value_counts())
```

```python
if "year" in df.columns:
    df = df.dropna(subset=["year"])
    display(df["year"].value_counts())
```

```python
if "year" in df.columns:
    df["year"] = pd.to_numeric(
        df["year"],
        errors="coerce"
    )

    df = df.dropna(subset=["year"])

    df["year"] = df["year"].astype(int)
    display(df["year"].value_counts())
```

```python
if "age" in df.columns:
    df["age_numeric"] = pd.to_numeric(
        df["age"],
        errors="coerce"
    )
    display(df["age_numeric"].value_counts())
```

```python
if "age_numeric" in df.columns:
    age_median = df["age_numeric"].median()

    df["age_numeric"] = (
        df["age_numeric"]
        .fillna(age_median)
        .astype(int)
    )
    display(df["age_numeric"].value_counts())
```

```python
numeric_columns = df.select_dtypes(
    include="number"
).columns

for col in numeric_columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

    df[col] = df[col].astype(int)
    display(df[col].value_counts())
```

```python
remaining_nulls = df.isnull().sum()

display(remaining_nulls[remaining_nulls > 0])
```
