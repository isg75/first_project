```python
import pandas as pd
import yaml



try:
    with open("../cfg.yaml", "r") as file:
        cfg=yaml.safe_load(file)
except:
    print("Yaml configuration file not found!")
```

```python
gdp=pd.read_csv(cfg['input_data']['df2'])
```

```python
# CRenaming columns for better undertanding and easier manipulation:
map_columns={'Country/Area':'country', 'Year':'year','GDP, at current prices - US Dollars':'GDP'}
gdp=gdp.rename(columns=map_columns)
gdp.columns
```

```python
# Dropping unnecesary columns:
gdp=gdp.drop(columns=['Unit'])
gdp.columns
```

```python
gdp.head(20)
```

```python
# Creating population df:
pop=pd.read_csv(cfg['input_data']['df3'])
pop.head()
```

```python
# Rename columns, drop unnecessary columns and check for duplicates or missing values:
map_popul={'Country/Area':'country', 'Year':'year','Population':'population'}
pop=pop.rename(columns=map_popul)
pop=pop.drop(columns=['Unit'])
display(pop.columns)
display(gdp.isna().sum())
display(gdp.duplicated().sum())
```

```python
display(gdp.shape)
display(pop.shape)
display(gdp.dtypes)
display(pop.dtypes)
```

```python
# Join population and gdp in a new dataframe:
gdp_pop = pd.merge(
    left=gdp, 
    right=pop,
    how='left',
    left_on=['country', 'year'],
    right_on=['country', 'year'],)

display(gdp_pop.head())
display(gdp_pop.shape)
display(gdp_pop.dtypes)

```

```python
# Handling missing values:
display(gdp_pop.duplicated().sum())
display(gdp_pop.isna().sum())
gdp_pop.loc[gdp_pop.isna().any(axis=1)]
```

```python
# Drop rows with missing values as they are irrelevant to the study (they belong to a region classification not a country):
gdp_pop=gdp_pop.dropna()
display(gdp_pop.isna().sum())
```

```python
# create a new column for gdp per capita:
gdp_pop['gdp_per_cap']=gdp_pop['GDP'] / gdp_pop['population']
display(gdp_pop.head())
```

```python
# New data set:

renew=pd.read_csv(cfg['input_data']['df1'])
```

```python
display(renew.head())
display(renew.info)
```

```python
renew.columns
```

```python
def clean_years(col):
    return col.split('[')[0].strip()

renew.columns=[clean_years(col) for col in renew.columns]
renew.columns
years=['2015','2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024']
```

```python
renew=renew.drop(columns=['Series Code','Country Code'])
renew.columns
```

```python
def rename_column(col):
    return col.lower().replace(' ','_').strip()

renew.columns=[rename_column(col) for col in renew.columns]
renew.columns
```

```python
renew.isna().sum()
```

```python
renew.duplicated().sum()
```

```python
renew.duplicated()
```

```python
renew.loc[[798,799]]
```

```python
renew = renew.drop(index=[798,799])
renew.duplicated().sum()
```

```python
renew.isna().sum()
```

```python
renew.loc[renew.isna().any(axis=1)]
```

```python
renew=renew.dropna()
display(renew.isna().sum())
```

```python
renew_melt = pd.melt(renew, id_vars=['country_name', 'series_name'], value_vars=years, var_name='year', value_name='value')
display(renew_melt.head())
```

```python
renew_df=renew_melt.pivot(index=['country_name','year'],columns='series_name',values='value').reset_index()
renew_df.head()
```

```python
renew_df.duplicated().sum()
```

```python
renew_df.isna().sum()
```

```python
# renaming new columns to practical names:
map_renew={'country_name':'country','Access to electricity (% of population)':'prct_access_elec','Electricity production from renewable sources, excluding hydroelectric (% of total)':'prct_renew_prod','Renewable energy consumption (% of total final energy consumption)':'prct_renew_cons'}
renew_df=renew_df.rename(columns=map_renew)
renew_df.columns
```

```python
# Save df to the output path defined in config, with reset index
renew_df.to_csv(cfg['output_data']['df1'], index=False)

print(f"File saved to: {cfg['output_data']['df1']}")
```
