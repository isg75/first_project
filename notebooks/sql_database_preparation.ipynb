```python
import pandas as pd
```

```python
sql_database = pd.read_csv("../data/clean/refined_dataset.csv", sep=';')
```

```python
sql_database.isna().sum()
```

```python
sql_database[sql_database["artist_name"].isna()]
sql_database[sql_database["track_name"].isna()]
sql_database[sql_database["album_name"].isna()]
```

```python
clean = sql_database.dropna(subset=["track_name", "artist_name", "album_name"]).copy()
```

```python
albums = clean[["album_name", "album_release_date"]].drop_duplicates().reset_index(drop=True)
albums["album_id"] = albums.index + 1
albums = albums[["album_id", "album_name", "album_release_date"]]
display(albums)
```

```python
genres = clean[["artist_genres"]].drop_duplicates().reset_index(drop=True)
genres["genre_id"] = genres.index + 1
genres = genres[["genre_id", "artist_genres"]]
display (genres)
```

```python
artist = clean[["artist_name", "artist_popularity", "artist_followers"]].drop_duplicates(subset="artist_name")
display(artist)
```

```python
tracks = clean.merge(albums, on=["album_name", "album_release_date"])
tracks = tracks[["track_id", "track_name", "track_popularity", "album_id"]]
display(tracks)
```

```python
track_artist = clean[["track_id", "artist_name"]]
display(track_artist)
```

```python
artist_genre = clean[["artist_genres", "artist_name"]].drop_duplicates()
artist_genre = artist_genre.merge(genres, on="artist_genres")
artist_genre = artist_genre[["genre_id", "artist_name"]]
display (artist_genre)
```
