```python
Background of the Study

Airbnb hosts often receive lower guest ratings without clearly understanding which listing characteristics contribute to guest dissatisfaction.

Since ratings directly influence booking decisions and listing visibility, identifying the features associated with poorly rated properties can help hosts prioritize improvements that enhance guest satisfaction and increase competitiveness.
```

```python
Objectives
General Objective:

To analyze Airbnb listing characteristics that influence guest ratings and identify practical strategies for improving underperforming listings.

Specific Objectives:
Compare the characteristics of poorly rated and highly rated Airbnb listings.
Determine whether pricing influences guest ratings.
Examine the relationship between property size and guest ratings.
Identify the amenities associated with highly rated listings.
Investigate whether location influences guest satisfaction.
Identify the strongest predictors of Airbnb ratings.
Develop business recommendations that can improve listing performance.
```

```python
Research Questions

RQ1. What is the distribution of Airbnb listings based on high, average, and low ratings?

RQ2. Is there a significant difference in price between poorly rated and highly rated Airbnb listings?

RQ3. Which property types are more likely to receive poor guest ratings?

RQ4. Do amenities significantly influence the price of Airbnb listings?

RQ5. Do amenities significantly influence Airbnb ratings?

RQ6. Does the location of an Airbnb listing significantly influence guest ratings?

RQ7. Does the location of an Airbnb listing significantly influence Airbnb prices?

RQ8. Does the number of reviews significantly influence Airbnb ratings?

RQ9. Which month has the highest number of reviews and the highest average Airbnb rating?

RQ10. What improvements should Airbnb hosts prioritize to transform poorly rated listings into highly rated accommodations?
```
