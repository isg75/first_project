```python
df.detail = (
    df.Detail
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
    .str.replace("/", "_", regex=False)
    .str.replace("*", "_", regex=False)
    .str.replace(r"[^a-z0-9_]", "", regex=True)
)
display (df)
```

```python
import pandas as pd

def title_extractor(df: pd.DataFrame) -> pd.DataFrame:
    temp_df = df.copy()
    temp_df[["Header", "State", "Country"]] = temp_df["Title"].str.split(
        ",", expand=True
    )
    temp_df.drop("Title", inplace=True, axis=1)
    return temp_df
```

```python
df_extracted = title_extractor(df)
display(df_extracted.head(30))
```

```python
df_extracted.columns = [(column.lower()).replace(" ","_") for column in df_extracted.columns]
display(df_extracted)
```

```python
df_extracted[['raiting','reviews']] = df_extracted.review_and_rating.str.split(" ", expand=True)
display(df_extracted)
```

```python
df_extracted["reviews"] = df_extracted["reviews"].str.replace(r'[()]', '', regex=True)
display(df_extracted)
```

```python
df_extracted.drop(['detail', 'date','number_of_bed','review_and_rating'], axis=1, inplace=True)
display(df_extracted)
#df_extracted.dropna(subset=['detail', 'date','review_and_rating','number_of_bed','raiting_and_reviews'])
```

```python
import numpy as np
import seaborn as sns
from statistics import mode
import matplotlib.pyplot as plt


season_country_df = df_extracted(['Clean_country','Season'])['Date'].agg(['count']).reset_index().pivot(index='Season', columns='Clean_country', values='count')

season_country_new_df = season_country_df.unstack("Season").reset_index(level=[0, 1]).copy()
season_order = {"Winter": 1, "Spring": 2, "Summer": 3, "Autum": 4}
s = season_country_new_df.sort_values(by=["Clean_country","Season"], key=lambda x: x.map(season_order))
s.columns = ["Country", "Season", "Number_of_Attacks"]
display(s)
sns.barplot(data=s, x="Season", y="Number_of_Attacks", hue="Country");
#plt.save_fig("season_country.png")
plt.show()
#season_country_df.plot(figsize=(8, 4))
#plt.show()
```
