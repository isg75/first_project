**All-time favorites**

```python
# All-time favorite tracks based on popularity and artist followers
track_data = last_decade_tracks.copy()
artist_data = last_decade_tracks.copy()

max_followers = track_data["artist_followers"].max()
track_data["follower_score"] = (
    track_data["artist_followers"] / max_followers * 100
)
track_data["favorite_score"] = (
    track_data["track_popularity"] * 0.8
    + track_data["follower_score"] * 0.2
)

top_overall = (
    track_data.sort_values("favorite_score", ascending=False)
    .drop_duplicates(subset="track_id")
)

# All-time Favorite Track: 2015 to 2018
nominee_1 = top_overall[
    top_overall["album_release_date"].between(2015, 2018)].head(1)

display(nominee_1)

# All-time Favorite Track: 2019 to 2022
nominee_2 = top_overall[
    top_overall["album_release_date"].between(2019, 2022)].head(1)

display(nominee_2)

# All-time Favorite Track: 2022 to 2025
nominee_3 = top_overall[
    top_overall["album_release_date"].between(2023, 2025)].head(1)

display(nominee_3)
```

```python
# All-Time Favorite Artist
overall_data = artist_data.pivot_table(
    index="artist_name",
    values=["track_popularity", "artist_followers"],
    aggfunc={
        "track_popularity": "median",
        "artist_followers": "max"
    }
).rename(columns={"track_popularity": "median_track_popularity"})

overall_data = overall_data.dropna(
    subset=["median_track_popularity", "artist_followers"]
).reset_index()

max_followers = overall_data["artist_followers"].max()
overall_data["follower_score"] = (
    overall_data["artist_followers"] / max_followers * 100
)

overall_data["favorite_score"] = (
    overall_data["median_track_popularity"] * 0.8
    + overall_data["follower_score"] * 0.2
)

top_overall_artists = (
    overall_data
    .sort_values("favorite_score", ascending=False)
    .reset_index(drop=True)
)

top_overall_artists.head(10)
```
