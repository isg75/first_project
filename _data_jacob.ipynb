```python
import pandas as pd
```

```python
df = pd.read_csv('track_data_final.csv')
```

```python
df.head()
```

```python
df["artist_genres"].isna().sum()
```

```python
df.shape
```

```python
df_clean = df.copy()
```

```python
df_clean = df_clean.drop(
    columns=["track_number", "explicit"]
)
```

```python
df_clean.head(10)
```

```python
df_clean["artist_genres"] = (
    df_clean["artist_genres"]
    .fillna("unknown")
    .str.lower()
    .str.strip()
)
```

```python
df_clean.columns
```

```python
df_clean.info()
```

```python
df_clean["artist_genres"] = (
    df_clean["artist_genres"].replace("[]","unknown")
)
```

```python
df_clean.head(10)
```

```python
df_clean["artist_genres"].isna().sum()
```

```python
df_clean["years"] = df_clean["album_release_date"].str[:4]
```

```python
# Albums have a higher average track popularity than single tracks.
df_clean.groupby("album_type")["track_popularity"].mean()
```

```python
#Artist popularity differs according to album type.
df_clean.groupby("album_type")["artist_popularity"].mean()
```

```python
album_type_popularity = (
    df_clean.groupby("album_type")["track_popularity"]
    .agg(["count", "mean", "median"])
    .round(2)
)

album_type_popularity
```

```python
# Artists releasing albums have more followers than artists releasing singles.
df_clean.groupby("album_type")["artist_followers"].median()
```
