```python
import pandas as pd
```

```python
df_final = pd.read_csv("ironhack_idol_df.csv")
df_final.head()
```

```python
spotify = df_final.copy()
```

```python
spotify.info()
```

```python
artist_data = spotify[
    [
        "artist_name",
        "artist_popularity",
        "artist_followers",
        "main_genre"
    ]
].drop_duplicates(subset="artist_name").reset_index(drop=True)
```

```python
artist_data.insert(
    0,
    "artist_id",
    range(1, len(artist_data) + 1)
)
```

```python
artist_data.head()
```

```python
spotify = spotify.merge(
    artist_data[["artist_id", "artist_name"]],
    on="artist_name",
    how="left"
)
```

```python
artist_data.to_csv("artist_data.csv", index=False, encoding="utf-8", lineterminator='\r\n')
```

```python
artist_data.shape
```

```python
album_data = spotify[
    [
        "album_id",
        "album_name",
        "album_type",
        "album_release_year"
    ]
].drop_duplicates(subset="album_id").reset_index(drop=True)
```

```python
album_data.shape
```

```python
album_data.head()
```

```python
album_data.to_csv("album_data.csv", index=False, encoding="utf-8", lineterminator='\r\n')
```

```python
track_data = spotify[
    [
        "track_id",
        "track_name",
        "track_popularity",
        "track_duration_min",
        "album_id",
        "artist_id"
    ]
].drop_duplicates(subset="track_id").reset_index(drop=True)
```

```python
track_data.to_csv("track_data.csv", index=False, encoding="utf-8", lineterminator='\r\n')
```
