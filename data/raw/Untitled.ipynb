```python
import pandas as pd

spotify_clean = pd.read_csv("data/raw/spotify_data_clean.csv")
spotify_track = pd.read_csv("data/raw/track_data_final.csv")

merge_spotify_data = pd.merge(spotify_clean, spotify_track, on="track_id", how="outer")

merge_spotify_data.to_csv("/Users/martaatencia/Desktop/Ironhack/Week_4/Day_1/first_project/data/raw/merged_spotify_data.csv", sep=";", encoding="utf-8")

display(merge_spotify_data)
```

```python
# Check if the track_id is a unique identifier
```

```python
unique_count = merge_spotify_data["track_id"].nunique()
print(unique_count)
```

```python
# The number of unique track_id values is the same as the total rows, then we can determine there aren't any duplicated records in the dataset.
```

```python
# Check the column names and data types
merge_spotify_data.info()
```

```python
# There are two columns for genres, however, one column has some empty values. We will use the records in the other column to fill the empty values.
```

```python
merge_spotify_data["artist_genres_y"] = merge_spotify_data["artist_genres_y"].str.strip("[]")

merge_spotify_data["artist_genres_y"] = merge_spotify_data["artist_genres_y"].str.strip("''")

merge_spotify_data.head()
```

```python
merge_spotify_data["artist_genres_x"] = merge_spotify_data["artist_genres_x"].fillna(merge_spotify_data["artist_genres_y"])

merge_spotify_data.head(10)
```

```python
# We do the same for the rest of the columns.
```

```python
merge_spotify_data["track_name_x"] = merge_spotify_data["track_name_x"].fillna(merge_spotify_data["track_name_y"])

merge_spotify_data["track_popularity_x"] = merge_spotify_data["track_popularity_x"].fillna(merge_spotify_data["track_popularity_y"])

merge_spotify_data["artist_name_x"] = merge_spotify_data["artist_name_x"].fillna(merge_spotify_data["artist_name_y"])

merge_spotify_data["artist_popularity_x"] = merge_spotify_data["artist_popularity_x"].fillna(merge_spotify_data["artist_popularity_y"])

merge_spotify_data["artist_followers_x"] = merge_spotify_data["artist_followers_x"].fillna(merge_spotify_data["artist_followers_y"])

merge_spotify_data["album_name_y"] = merge_spotify_data["album_name_y"].fillna(merge_spotify_data["album_name_x"])

merge_spotify_data["album_release_date_y"] = merge_spotify_data["album_release_date_y"].fillna(merge_spotify_data["album_release_date_x"])

merge_spotify_data.head(10)

```

```python
merge_spotify_data = merge_spotify_data.drop(columns = ["artist_genres_y"])

merge_spotify_data.info()
```

```python
# We filter out unnecessary columns
refined_dataset = merge_spotify_data[["track_id", "track_name_x", "track_popularity_x", "artist_name_x", "artist_popularity_x", "artist_followers_x", "artist_genres_x", "album_name_y", "album_release_date_y"]]
display(refined_dataset)
```

```python
# We standardized the column names
```

```python
refined_dataset.columns = (refined_dataset.columns
    .str.replace("_x","")
    .str.replace("_y","")
)
refined_dataset.info()
```

```python
# First, we check if there are any records with empty values for date. Then, we clean date column to keep only the year.
```

```python
refined_dataset["album_release_date"].isnull().sum()
```

```python
refined_dataset["album_release_date"] = refined_dataset["album_release_date"].astype(str).str[:4]
```

```python
unique_dates = refined_dataset["album_release_date"].unique()
print(unique_dates)
```

```python
unique_date_count = refined_dataset["album_release_date"].value_counts()
unique_date_count.head(40)
```

```python
# We create a new table with records from the last 10 years (2015-2025). To filter by date, we need to first change the data type from string to integer.
```

```python
refined_dataset["album_release_date"] = refined_dataset["album_release_date"].astype("int")
refined_dataset.info()
```

```python
last_decade_tracks = refined_dataset[refined_dataset["album_release_date"] > 2014]
display(last_decade_tracks)
```
